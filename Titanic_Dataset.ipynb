```python

# ============================================
# CHART SELECTION AND GRAPHICAL PERCEPTION
# Titanic Dataset - Kaggle
# ============================================

# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Upload the original Kaggle dataset
from google.colab import files

uploaded = files.upload()

# Step 3: Read the actual dataset
df = pd.read_csv("train.csv")

# Step 4: Display dataset information
print("Dataset Information:")
print(df.info())

print("\nFirst five rows:")
display(df.head())

# Step 5: Calculate survival counts
survival_counts = df["Survived"].value_counts().sort_index()

# Rename categories
labels = ["Did not survive", "Survived"]
counts = [
    survival_counts.get(0, 0),
    survival_counts.get(1, 0)
]

total = sum(counts)
percentages = [count / total * 100 for count in counts]

print("\nSurvival Summary:")
for label, count, percentage in zip(labels, counts, percentages):
    print(f"{label}: {count} ({percentage:.1f}%)")

# ============================================
# Step 6: Create Bar Chart
# ============================================

plt.figure(figsize=(8, 5))

bars = plt.bar(labels, counts)

plt.title("Titanic Passenger Survival - Bar Chart")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.ylim(0, max(counts) * 1.2)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 5,
        f"{int(height)}",
        ha="center"
    )

plt.tight_layout()
plt.savefig("titanic_bar_chart.png", dpi=300)
plt.show()

# ============================================
# Step 7: Create Pie Chart
# ============================================

plt.figure(figsize=(7, 7))

plt.pie(
    counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Titanic Passenger Survival - Pie Chart")
plt.tight_layout()
plt.savefig("titanic_pie_chart.png", dpi=300)
plt.show()

# ============================================
# Step 8: Download Both Charts
# ============================================

files.download("titanic_bar_chart.png")
files.download("titanic_pie_chart.png")

print("\nBoth charts were successfully generated.")
```
